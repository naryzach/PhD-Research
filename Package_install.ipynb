Create local folders used by scripts

```python
from pathlib import Path
import os

directory_names = ["Local/", "Tools/"]
for dir_name in directory_names:
    dir_path = Path(dir_name)
    dir_path.mkdir(parents=True, exist_ok=True)
    print(f"Directory '{dir_path}' created (or already existed).")

file_path = "credentials.ini"
if not os.path.exists(file_path):
    with open(file_path, 'w') as f:
        print(f"Created empty credentials file: {file_path}")
else:
    print(f"Credentials file already exists: {file_path}")
```
