```python
import os
import subprocess
import json
```

```python
# --- Path to ProteinMPNN directory ---
bindcraft_repo = "../Tools/BindCraft"

# --- Input PDB file ---
data_path = "../Data"
pdb_path = os.path.join(data_path, "TIMP3_vs_ADAM17_X_ray.pdb")
pdb_basename = os.path.basename(pdb_path).replace(".pdb", "")

# --- Output directory (will be created if not exists) ---
output_dir = "../Local/BindCraft_Designs"
settings_target = os.path.join(output_dir, "settings_target.json")
settings_filters = os.path.join(output_dir, "settings_filters.json")
settings_advanced = os.path.join(output_dir, "settings_advanced.json")

# --- Chains and residues to redesign ---
chains_to_design = ["B"]
residues_to_redesign = [31, 32, 33, 34, 35, 36]  # specify the residues to mutate

original_directory = os.getcwd()
```

```python
# Ensure output directory exists
os.makedirs(output_dir, exist_ok=True)

# setup settings
bc_target = {}
bc_target["design_path"] = "/content/drive/My Drive/BindCraft/PDL1/"
bc_target["binder_name"] = "PDL1"
bc_target["starting_pdb"] = "../Data/PDL1.pdb"
bc_target["chains"] = "A"
bc_target["target_hotspot_residues"] = "56"
bc_target["lengths"] = [65, 150]
bc_target["number_of_final_designs"] = 100

with open(settings_target, 'w') as f:
    json.dump(bc_target, f, indent=2)
```

```python
fix_command = [
    "python",
    "-u", bindcraft_repo.replace("../", "") + "/bindcraft.py",
    "--settings", settings_target,
    "--filters", bindcraft_repo.replace("../", "") + '/settings_filters/default_filters.json',
    #"--advanced", bindcraft_repo.replace("../", "") + '/settings_advanced/default_4stage_multimer.json'
]

print(" ".join(fix_command))

os.chdir("..")
result = subprocess.run(fix_command, capture_output=True, text=True)
os.chdir(original_directory)

print(result.stdout)
print(result.stderr)
```
